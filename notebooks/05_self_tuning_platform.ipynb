# 自调优训练平台 (PBT)

`PBTController` 用 Population-Based Training 自动搜索超参。

每代会：
- 评估一整群 trial
- 淘汰差的，用精英的配置 + 扰动产生新 trial


````python
from arcagent.config import RunConfig
from arcagent.platform import PBTController
import inspect

print(inspect.getsource(PBTController.__init__)[:700])
````

## 跑一个小 PBT 演示（非常短）

````python
# 注意：真实运行用 CLI 或完整 steps。这里只是展示 API
cfg = RunConfig(run_dir="runs/notebook_pbt_demo", train={"steps": 50})
pbt = PBTController(population=3, generations=1, steps_per_generation=50, base_config=cfg)
print("PBTController ready. population size:", len(pbt.trials))
# pbt.run()  # 取消注释可真正跑（会花一些时间）
````

完整使用看项目根目录的 CLI：
```bash
python -m arcagent.platform.pbt --generations 2 --population 4 --steps-per-generation 200
```
