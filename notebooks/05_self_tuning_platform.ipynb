# 自调优训练平台

平台会同时训练多个 trial。每一代结束后，它保留表现好的配置，把差的 trial 替换成好配置的轻微变体。这个过程叫 Population-Based Training。

```python
from arcagent.config import RunConfig
from arcagent.platform import PBTController
cfg = RunConfig(run_dir='runs/notebook_pbt')
cfg.platform.population = 2
cfg.platform.generations = 1
cfg.platform.steps_per_generation = 50
states = PBTController(cfg).run()
[(s.trial_id, s.score) for s in states]
```
