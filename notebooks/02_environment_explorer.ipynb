# 环境探索：agent 看到什么？

这个 notebook 展示 ARC 风格网格、动作空间和一步交互。

```python
from arcagent.envs import MockArcEnv
env = MockArcEnv(grid_size=16, max_steps=20, seed=0)
obs = env.reset()
print('grid shape:', obs['grid'].shape)
print('actions:', obs['available_actions'])
print(obs['grid'])
```

颜色 2/6 是玩家，4 是钥匙，3 是目标，8 是墙。先拿钥匙再到目标，是一个多步依赖。

```python
for action in [4,4,4,2,2]:
    obs, reward, done, info = env.step(action)
    print({'action': action, 'reward': round(reward, 3), 'done': done, 'info': info})
```
