# 环境探索：Agent 看到什么、能做什么？

本 notebook 带你彻底搞懂 `arcagent.envs` 层。

核心类：
- `MockArcEnv`（本地主要使用的模拟环境）
- `ArcEngineEnv`（尝试使用官方 arc_agi toolkit 的适配器）
- Action 编码/解码在 `arcagent.envs.actions`


## 1. 基本用法

```python
from arcagent.envs import MockArcEnv
import numpy as np

env = MockArcEnv(grid_size=12, max_steps=50, seed=123)
obs = env.reset()
print("初始 obs keys:", list(obs.keys()))
print("grid dtype:", obs["grid"].dtype, "shape:", obs["grid"].shape)
print("available_actions:", obs["available_actions"])
```

## 2. 直接查看环境源码

```python
import inspect
from arcagent.envs.mock_arc import MockArcEnv as MAE
print(inspect.getsource(MAE.step)[:900])
```

## 3. Action 空间

```python
from arcagent.envs import actions
print("Action space size:", actions.ACTION_SPACE_SIZE)
print("Action names (first 8):", list(actions.ACTION_NAMES.items())[:8])
print("Example encode/decode:")
print(actions.encode_action(0))
```

## 4. 练习
1. 把 grid_size 改成 8 或 24，重置后看 grid 形状。
2. 手动调用 step 多次，观察 done 和 reward。
3. 阅读 `arcagent/envs/actions.py` 完整实现。


```python
# 你的实验区
obs = env.reset()
for i in range(5):
    a = np.random.choice(obs["available_actions"])
    obs, reward, done, info = env.step(a)
    print(f"step {i}: action={a}, reward={reward}, done={done}")
    if done: break
```
