这里演示如何加载本地 checkpoint 并用离线 submission shim 选择动作。

```python
from arcagent.envs import MockArcEnv
from submission.agent import SubmissionAgent
from submission.offline_check import scan
print('offline violations:', scan('.'))
env = MockArcEnv()
obs = env.reset()
agent = SubmissionAgent(grid_size=16)
action = agent.choose_action(obs)
print(action)
```
