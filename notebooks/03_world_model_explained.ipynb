# 世界模型：为什么叫“会想象”

`DreamerLiteWorldModel` 是整个项目的灵魂。它做三件事：
1. `encode(grid)` → latent
2. `predict_next_latent(latent, action)` → 想象下一步
3. `decode(latent)` → 还原成 grid + 预测 reward

训练目标：让预测的下一帧和真实下一帧尽量接近，同时学会哪个动作好。


## 实例化并查看实现

```python
from arcagent.models.dreamer_lite import DreamerLiteWorldModel
import inspect
import numpy as np

wm = DreamerLiteWorldModel(grid_size=16, latent_dim=32, seed=0)
print("Model created")
print(inspect.getsource(wm.encode)[:550])
```

## 手动跑 encode → predict → decode

```python
grid = np.random.randint(0, 16, (16, 16), dtype=np.int8)
latent = wm.encode(grid)
print("latent shape:", latent.shape)

next_latent = wm.predict_next_latent(latent, action=3)
print("next latent shape:", next_latent.shape)

pred_grid, pred_reward = wm.decode(next_latent)
print("pred grid shape:", pred_grid.shape, "pred reward:", pred_reward)
```

## 看训练逻辑 (train_batch)

```python
print(inspect.getsource(wm.train_batch)[:900])
```

## 关键洞见
这个世界模型非常小（纯 NumPy + 手动 SGD），但完整实现了 Dreamer 风格的 latent dynamics + reward prediction。
