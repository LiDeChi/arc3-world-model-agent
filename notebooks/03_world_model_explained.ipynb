# 世界模型：为什么叫“会想象”

世界模型做三件事：encode(grid) 把网格压缩成 latent；predict_next_latent(latent, action) 预测动作后的 latent；decode(latent) 把想象状态还原成网格。

```python
from arcagent.agent import WorldModelAgent
from arcagent.envs import MockArcEnv
env = MockArcEnv(); obs = env.reset()
agent = WorldModelAgent.create(grid_size=16, seed=0)
imagined = agent.imagined_rollout(obs['grid'], steps=4)
[g.shape for g in imagined]
```

训练时，真实下一帧会监督模型的预测下一帧；奖励头会学习哪个动作更可能带来好结果。
