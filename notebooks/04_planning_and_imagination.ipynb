# 规划与想象

agent 不只试一个动作，而是在世界模型里对多个未来动作序列打分。

```python
from arcagent.agent import WorldModelAgent
from arcagent.envs import MockArcEnv
env = MockArcEnv(seed=2)
obs = env.reset()
agent = WorldModelAgent.create(grid_size=16, seed=2, epsilon=0.0)
for action in [1,2,3,4,5]:
    score = agent._score_first_action(obs['grid'], action, [1,2,3,4,5])
    print(action, round(score, 4))
print('chosen:', agent.act(obs, explore=False))
```

未训练模型一开始想象很粗糙；训练后，预测奖励和下一帧会逐渐更接近真实环境。
