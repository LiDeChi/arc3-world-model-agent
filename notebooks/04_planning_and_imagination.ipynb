# 规划与想象 (Planning with Imagination)

这是 agent 决策的核心：**在头脑里 rollout 多条未来，然后挑最好的一条**。

关键方法：`WorldModelAgent.act(..., explore=False)`


```python
from arcagent.agent import WorldModelAgent
from arcagent.envs import MockArcEnv
import inspect

agent = WorldModelAgent.create(grid_size=16, imagination_horizon=5, planning_rollouts=8, seed=99)
env = MockArcEnv(grid_size=16, seed=99)
obs = env.reset()

print(inspect.getsource(agent.act)[:1100])
```

## 实际调用一次决策

```python
action = agent.act(obs["grid"], explore=False)
print("Chosen action:", action)

# 看它想象了什么（简化版）
imagined = agent.imagined_rollout(obs["grid"], steps=agent.config.imagination_horizon)
print("Number of imagined steps:", len(imagined))
```
